```python
import asyncio
import aiohttp
```

```python
async def get_data(id: int, endpoint: str):
    print(f"Started: {id}")
    url = f"http://127.0.0.1:8000/{endpoint}/{id}"
    async with aiohttp.ClientSession() as session:
        async with session.get(url):
            print(f"Ended: {id}")
```

```python
await get_data(1, "async")
```

```python
await asyncio.gather(*[get_data(i, "async") for i in range(100)])
```
